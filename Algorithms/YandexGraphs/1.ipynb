```python
# import sys


def main():
    """
    Пример ввода и вывода числа n, где -10^9 < n < 10^9:
    n = int(input())
    print(n)
    """

    x = [int(i) for i in input().split()]
    n = len(x)
    ans = [0] if x[0] != 0 else []
    for i in range(n - 1):
        x = [int(j) for j in input().split()]
        if x[i + 1] != 0:
            ans.append(i + 1)
    if len(ans) == 0:
        print('NO LOOPS')
    else:
        for i in ans:
            print(i)
    pass


if __name__ == '__main__':
    main()
```
