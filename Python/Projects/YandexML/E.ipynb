```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import LabelEncoder

# Assume df_train and df_test are already loaded in your environment

# 1. Concatenate for consistent preprocessing
# We tag them to separate them later
df_train['is_train'] = 1
df_test['is_train'] = 0
df_test['target'] = np.nan # Placeholder

# Combine
full_data = pd.concat([df_train, df_test], axis=0).reset_index(drop=True)

# 2. Define ID columns to drop from features
# snapshot_dt might contain trend info, but usually, we drop it for basic classification
id_cols = ['agreement', 'customer', 'snapshot_dt', 'is_train', 'target']
features = [c for c in full_data.columns if c not in id_cols]

# 3. Handle Categorical Features (Object columns)
# We use Label Encoding: turning strings into integers. 
# XGBoost handles integers well even if they represent categories.
object_cols = full_data.select_dtypes(include=['object']).columns

print(f"Encoding {len(object_cols)} categorical columns...")

for col in object_cols:
    le = LabelEncoder()
    # Convert to string to handle mixed types/NaNs uniformly for the encoder
    full_data[col] = full_data[col].astype(str)
    full_data[col] = le.fit_transform(full_data[col])

# 4. Split back into Train and Test
X = full_data[full_data['is_train'] == 1][features]
y = full_data[full_data['is_train'] == 1]['target']
X_test = full_data[full_data['is_train'] == 0][features]

# 5. Model Training with Cross-Validation
# We use K-Fold to ensure the model is stable and not overfitting
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X))
test_preds = np.zeros(len(X_test))

print("Starting Training...")

for fold, (train_idx, val_idx) in enumerate(kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    # XGBoost configuration
    model = xgb.XGBClassifier(
        n_estimators=1000,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        n_jobs=-1,
        random_state=42,
        eval_metric='auc',
        early_stopping_rounds=50,
        # 'enable_categorical': True # Uncomment if using very recent XGBoost versions
    )
    
    model.fit(
        X_train, y_train,
        eval_set=[(X_val, y_val)],
        verbose=False
    )
    
    # Predict validation set (to check accuracy)
    val_pred = model.predict_proba(X_val)[:, 1]
    oof_preds[val_idx] = val_pred
    
    # Predict on actual test set
    test_preds += model.predict_proba(X_test)[:, 1] / kf.get_n_splits()
    
    score = roc_auc_score(y_val, val_pred)
    print(f"Fold {fold+1} AUC: {score:.4f}")

# 6. Final Evaluation
overall_auc = roc_auc_score(y, oof_preds)
print(f"--- Overall CV AUC: {overall_auc:.4f} ---")



# 7. Create Submission DataFrame
submission = pd.DataFrame({
    'customer': df_test['customer'], # Assuming customer ID is required for submission
    'target': test_preds
})

# submission.to_csv('submission.csv', index=False)
print("Prediction complete. Head of submission:")
print(submission.head())
```

```python
# Вариант B: подбор порога по OOF (максимизируем F1)
from sklearn.metrics import precision_recall_curve, f1_score, accuracy_score

precision, recall, thresholds = precision_recall_curve(y, oof_preds)
# precision_recall_curve возвращает thresholds длиной len(precision)-1
f1_scores = 2 * precision[:-1] * recall[:-1] / (precision[:-1] + recall[:-1] + 1e-12)
best_idx = f1_scores.argmax()
best_thr = thresholds[best_idx]

print(f"Лучший порог (по F1 на OOF): {best_thr:.4f}")
print(f"Лучший F1 на OOF: {f1_scores[best_idx]:.4f}")

test_pred_binary = (test_preds >= best_thr).astype(int)
oof_pred_binary = (oof_preds >= best_thr).astype(int)

print(f"F1 (best_thr) on OOF: {f1_score(y, oof_pred_binary):.4f}")
print(f"Accuracy (best_thr) on OOF: {accuracy_score(y, oof_pred_binary):.4f}")
```

```python
# --- Optuna + xgb.train (совместимо с версиями XGBoost, где sklearn.fit не принимает callbacks) ---
import optuna
from optuna.integration import XGBoostPruningCallback
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
import numpy as np
import pandas as pd

# Функция-цель для Optuna
def objective(trial):
    # пространство поиска
    params = {
        'objective': 'binary:logistic',
        'eval_metric': 'auc',
        'learning_rate': trial.suggest_loguniform('learning_rate', 0.01, 0.3),
        'max_depth': int(trial.suggest_int('max_depth', 3, 10)),
        'subsample': trial.suggest_uniform('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_uniform('colsample_bytree', 0.4, 1.0),
        'alpha': trial.suggest_loguniform('reg_alpha', 1e-8, 10.0),   # L1
        'lambda': trial.suggest_loguniform('reg_lambda', 1e-8, 10.0), # L2
        'min_child_weight': int(trial.suggest_int('min_child_weight', 1, 20)),
        'verbosity': 0,
        'seed': 42,
        # 'tree_method': 'hist',   # можете раскомментировать при больших датасетах или GPU 'gpu_hist'
    }

    kf_inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    scores = []

    for tr_idx, val_idx in kf_inner.split(X, y):
        X_tr, X_val = X.iloc[tr_idx], X.iloc[val_idx]
        y_tr, y_val = y.iloc[tr_idx], y.iloc[val_idx]

        dtrain = xgb.DMatrix(X_tr, label=y_tr)
        dval = xgb.DMatrix(X_val, label=y_val)

        evals = [(dval, 'valid')]

        # Добавляем pruning callback — если Optuna поддерживает для вашей версии XGBoost
        pruning_cb = XGBoostPruningCallback(trial, "valid-auc")

        # Тренируем через xgb.train — поддерживает early_stopping_rounds
        bst = xgb.train(
            params,
            dtrain,
            num_boost_round=2000,
            evals=evals,
            early_stopping_rounds=100,
            verbose_eval=False,
            callbacks=[pruning_cb] if pruning_cb is not None else None
        )

        # Предсказания на валидации: используем оптимальное количество деревьев, если есть
        if hasattr(bst, 'best_ntree_limit'):
            val_pred = bst.predict(dval, ntree_limit=bst.best_ntree_limit)
        else:
            val_pred = bst.predict(dval)

        scores.append(roc_auc_score(y_val, val_pred))

    return float(np.mean(scores))


# Создаём study
sampler = TPESampler(seed=42)
pruner = MedianPruner(n_startup_trials=5, n_warmup_steps=10)
study = optuna.create_study(direction='maximize', sampler=sampler, pruner=pruner)

N_TRIALS = 50  # подберите под ресурсы
print("Запуск Optuna-оптимизации...")
study.optimize(objective, n_trials=N_TRIALS, show_progress_bar=True)

print("Лучший trial:")
print(f"  CV AUC = {study.best_value:.5f}")
print("  Params:")
for k, v in study.best_params.items():
    print(f"    {k}: {v}")

# --- Финальное обучение (5-fold CV), используя xgb.train для каждого фолда ---
best = study.best_params
final_params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc',
    'learning_rate': best['learning_rate'],
    'max_depth': int(best['max_depth']),
    'subsample': best['subsample'],
    'colsample_bytree': best['colsample_bytree'],
    'alpha': best['reg_alpha'],
    'lambda': best['reg_lambda'],
    'min_child_weight': int(best['min_child_weight']),
    'verbosity': 0,
    'seed': 42,
}

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X))
test_preds = np.zeros(len(X_test))

dtest = xgb.DMatrix(X_test)

print("Тренируем финальные модели и собираем OOF + предсказания на тесте...")
for fold, (train_idx, val_idx) in enumerate(kf.split(X, y)):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]

    dtrain = xgb.DMatrix(X_tr, label=y_tr)
    dval = xgb.DMatrix(X_val, label=y_val)

    bst = xgb.train(
        final_params,
        dtrain,
        num_boost_round=2000,
        evals=[(dval, 'valid')],
        early_stopping_rounds=100,
        verbose_eval=False
    )

    # val predictions
    if hasattr(bst, 'best_ntree_limit'):
        val_pred = bst.predict(dval, ntree_limit=bst.best_ntree_limit)
        test_pred = bst.predict(dtest, ntree_limit=bst.best_ntree_limit)
    else:
        val_pred = bst.predict(dval)
        test_pred = bst.predict(dtest)

    oof_preds[val_idx] = val_pred
    test_preds += test_pred / kf.get_n_splits()

    fold_score = roc_auc_score(y_val, val_pred)
    print(f"Fold {fold+1} AUC: {fold_score:.4f}")

overall_auc = roc_auc_score(y, oof_preds)
print(f"--- Overall CV AUC (final): {overall_auc:.4f} ---")

submission = pd.DataFrame({
    'customer': df_test['customer'],
    'target': test_preds
})
print("Head of submission:")
print(submission.head())
# --- Конец блока ---
```

```python
import optuna
from optuna.integration import XGBoostPruningCallback
from optuna.samplers import TPESampler
from optuna.pruners import MedianPruner
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import warnings

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')

# ----------------------------
# 1. Config & Data Setup
# ----------------------------
target_col = 'target'       # Name of target variable
id_col = 'customer'         # ID column (will be dropped from features)

# Assumption: df_train and df_test are already loaded in memory
df_train = pd.read_csv('train.csv')
df_test = pd.read_csv('test.csv')

n_train = len(df_train)
train_with_target = df_train.copy()
test_df = df_test.copy()

# Concatenate for consistent encoding
full_data = pd.concat([train_with_target.drop(columns=[target_col], errors='ignore'),
                       test_df],
                      axis=0).reset_index(drop=True)

# ----------------------------
# 2. Basic Feature Engineering
# ----------------------------
# A. Row-wise null count
full_data['null_counts'] = full_data.isnull().sum(axis=1)

# B. Frequency encoding (for object/category)
object_cols = full_data.select_dtypes(include=['object', 'category']).columns.tolist()
for col in object_cols:
    full_data[f'{col}_freq'] = full_data[col].map(full_data[col].value_counts())

# C. Label encoding (safe conversion to int)
for col in object_cols:
    le = LabelEncoder()
    # Fill NaN with string to prevent LabelEncoder error
    full_data[col] = full_data[col].astype(str).fillna('##MISSING##')
    full_data[col] = le.fit_transform(full_data[col])

# Split back into X and X_test
X = full_data.iloc[:n_train, :].reset_index(drop=True)
X_test = full_data.iloc[n_train:, :].reset_index(drop=True)

# Extract Target
y = df_train[target_col].reset_index(drop=True)

# Drop ID columns
if id_col in X.columns:
    X = X.drop(columns=[id_col])
if id_col in X_test.columns:
    X_test = X_test.drop(columns=[id_col])

# ----------------------------
# 3. Class Imbalance (scale_pos_weight)
# ----------------------------
pos = float((y == 1).sum())
neg = float((y == 0).sum())
scale_pos_weight = neg / pos if pos > 0 else 1.0
print(f"Calculated scale_pos_weight: {scale_pos_weight:.2f}")

# ----------------------------
# 4. Optuna Objective with Target Encoding inside CV
# ----------------------------
def objective(trial):
    # UPDATED: Using suggest_float with log=True replaces suggest_loguniform
    params = {
        'objective': 'binary:logistic',
        'eval_metric': 'auc',
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.4, 1.0),
        'alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'lambda': trial.suggest_float('reg_lambda', 1e-8, 10.0, log=True),
        'min_child_weight': trial.suggest_int('min_child_weight', 1, 20),
        'verbosity': 0,
        'seed': 42,
        'scale_pos_weight': scale_pos_weight,
    }

    kf_inner = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    scores = []

    for tr_idx, val_idx in kf_inner.split(X, y):
        X_tr = X.iloc[tr_idx].copy().reset_index(drop=True)
        X_val = X.iloc[val_idx].copy().reset_index(drop=True)
        y_tr = y.iloc[tr_idx].reset_index(drop=True)
        y_val = y.iloc[val_idx].reset_index(drop=True)

        # === Target Encoding Inside Fold ===
        # Select high cardinality features
        cat_features = [c for c in X_tr.columns
                        if ('freq' not in c and 'null' not in c and c != id_col and X_tr[c].nunique() > 10)]

        smoothing = 10
        for col in cat_features:
            # Calculate stats on TRAIN ONLY
            temp = pd.DataFrame({col: X_tr[col], target_col: y_tr})
            agg = temp.groupby(col)[target_col].agg(['count', 'mean'])
            counts = agg['count']
            means = agg['mean']
            global_mean = y_tr.mean()

            # Smooth
            smooth = (counts * means + smoothing * global_mean) / (counts + smoothing)
            smooth_map = smooth.to_dict()

            # Map to train and val
            X_tr[f'{col}_target_enc'] = X_tr[col].map(smooth_map)
            X_val[f'{col}_target_enc'] = X_val[col].map(smooth_map)

            # Fill unseen values with global mean
            X_tr[f'{col}_target_enc'].fillna(global_mean, inplace=True)
            X_val[f'{col}_target_enc'].fillna(global_mean, inplace=True)

        # === Training ===
        dtrain = xgb.DMatrix(X_tr, label=y_tr)
        dval = xgb.DMatrix(X_val, label=y_val)

        pruning_cb = XGBoostPruningCallback(trial, "valid-auc")

        bst = xgb.train(
            params,
            dtrain,
            num_boost_round=2000,
            evals=[(dval, 'valid')],
            early_stopping_rounds=100,
            verbose_eval=False,
            callbacks=[pruning_cb]
        )

        # UPDATED: Use iteration_range instead of ntree_limit
        best_iteration = bst.best_iteration
        val_pred = bst.predict(dval, iteration_range=(0, best_iteration + 1))

        scores.append(roc_auc_score(y_val, val_pred))

    return np.mean(scores)

# ----------------------------
# 5. Run Optimization
# ----------------------------
sampler = TPESampler(seed=42)
pruner = MedianPruner(n_startup_trials=5, n_warmup_steps=10)
study = optuna.create_study(direction='maximize', sampler=sampler, pruner=pruner)

N_TRIALS = 50
print(f"Starting Optuna optimization for {N_TRIALS} trials...")
study.optimize(objective, n_trials=N_TRIALS, show_progress_bar=True)

print("\nBest Trial:")
print(f"  CV AUC = {study.best_value:.5f}")
print("  Params:")
for k, v in study.best_params.items():
    print(f"    {k}: {v}")

# ----------------------------
# 6. Final Training & OOF Generation
# ----------------------------
best = study.best_params
final_params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc',
    'learning_rate': best['learning_rate'],
    'max_depth': int(best['max_depth']),
    'subsample': best['subsample'],
    'colsample_bytree': best['colsample_bytree'],
    'alpha': best['reg_alpha'],
    'lambda': best['reg_lambda'],
    'min_child_weight': int(best['min_child_weight']),
    'verbosity': 0,
    'seed': 42,
    'scale_pos_weight': scale_pos_weight,
}

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(X))
test_preds = np.zeros(len(X_test))

print("\nTraining final models with best params...")

for fold, (train_idx, val_idx) in enumerate(kf.split(X, y)):
    X_tr = X.iloc[train_idx].copy().reset_index(drop=True)
    X_val = X.iloc[val_idx].copy().reset_index(drop=True)
    y_tr = y.iloc[train_idx].reset_index(drop=True)
    y_val = y.iloc[val_idx].reset_index(drop=True)

    # === Target Encoding (Exact same logic as CV) ===
    cat_features = [c for c in X_tr.columns
                    if ('freq' not in c and 'null' not in c and c != id_col and X_tr[c].nunique() > 10)]
    
    # Copy X_test for this fold to apply mapping safely
    X_test_fold = X_test.copy().reset_index(drop=True)
    smoothing = 10

    for col in cat_features:
        temp = pd.DataFrame({col: X_tr[col], target_col: y_tr})
        agg = temp.groupby(col)[target_col].agg(['count', 'mean'])
        counts = agg['count']
        means = agg['mean']
        global_mean = y_tr.mean()

        smooth = (counts * means + smoothing * global_mean) / (counts + smoothing)
        smooth_map = smooth.to_dict()

        X_tr[f'{col}_target_enc'] = X_tr[col].map(smooth_map)
        X_val[f'{col}_target_enc'] = X_val[col].map(smooth_map)
        # Apply Train map to Test
        X_test_fold[f'{col}_target_enc'] = X_test_fold[col].map(smooth_map)

        # Fill NaNs
        X_tr[f'{col}_target_enc'].fillna(global_mean, inplace=True)
        X_val[f'{col}_target_enc'].fillna(global_mean, inplace=True)
        X_test_fold[f'{col}_target_enc'].fillna(global_mean, inplace=True)

    # === Train ===
    dtrain = xgb.DMatrix(X_tr, label=y_tr)
    dval = xgb.DMatrix(X_val, label=y_val)
    dtest = xgb.DMatrix(X_test_fold)

    bst = xgb.train(
        final_params,
        dtrain,
        num_boost_round=2000,
        evals=[(dval, 'valid')],
        early_stopping_rounds=100,
        verbose_eval=False
    )

    # UPDATED: iteration_range
    best_iteration = bst.best_iteration
    val_pred = bst.predict(dval, iteration_range=(0, best_iteration + 1))
    test_pred = bst.predict(dtest, iteration_range=(0, best_iteration + 1))

    oof_preds[val_idx] = val_pred
    test_preds += test_pred / kf.get_n_splits()

    fold_score = roc_auc_score(y_val, val_pred)
    print(f"Fold {fold+1} AUC: {fold_score:.4f}")

overall_auc = roc_auc_score(y, oof_preds)
print(f"--- Overall CV AUC (final): {overall_auc:.4f} ---")

# ----------------------------
# 7. Submission
# ----------------------------
submission = pd.DataFrame({
    id_col: df_test[id_col],
    'target': test_preds
})
print("Head of submission:")
print(submission.head())
```
