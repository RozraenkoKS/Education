Вообще, нужно учитывать МПГ для величин из выборки с нормально распределёнными СВ, например, равенства средних какому либо одному, например, их среднему арифметическому, для похожих. Но p_value настолько мало, что это можно не учитывать. Однако, ниже реализация.

```python
def wald_test(sample, theta, estimation_theta, estimation_sigma, alternative='two_sided', alpha=0.05):
    """
    param sample: реализация выборки
    param theta: истинное значение параметра
    param estimation_theta: оценка параметра
    param estimation_sigma: оценка асимптотической дисперсии оценки estimation_sigma
    param alternative: вид альтернативной гипотезы, может принимать одно из значений 'two_sided', 'less', 'greater'

    return statistic
    return p_value
    """

    z = sps.norm.ppf(1 - alpha/2)
    n = len(sample)
    statistic = np.sqrt(n) * (estimation_theta - theta) / estimation_sigma

    if alternative == 'two_sided':
        p_value = sps.norm.sf(np.abs(statistic)) + sps.norm.cdf(-np.abs(statistic))
        conf_int = round(estimation_theta - z*estimation_sigma/np.sqrt(n), 4), round(estimation_theta + z*estimation_sigma/np.sqrt(n), 4)


    elif alternative == 'less':
        p_value = sps.norm.cdf(statistic)
        conf_int = (-np.inf, round(estimation_theta + z*estimation_sigma/np.sqrt(n), 4))

    
    elif alternative == 'greater':
        p_value = sps.norm.sf(statistic)
        conf_int = (round(estimation_theta + z*estimation_sigma/np.sqrt(n), 4), np.inf)

    else:
        raise ValueError('alternative name is wrong')

    return statistic, p_value, conf_int
```

```python
def estimation(sample):
    estimation_theta = np.mean(sample)
    estimation_sigma = np.sqrt(1 / (len(sample) - 1) * np.sum((sample - estimation_theta) ** 2))
    return estimation_theta, estimation_sigma
```

```python
sample1 = data[data['id'] == 321]['theta']
sample2 = data[data['id'] == 2212]['theta']

estimation_theta1, estimation_sigma1 = estimation(sample1)
estimation_theta2, estimation_sigma2 = estimation(sample2)

theta = (estimation_theta1 + estimation_theta2) / 2
alpha = 0.05

_1, p_value1, conf_int1 = wald_test(sample1, theta, estimation_theta1, estimation_sigma1)
_2, p_value2, conf_int2 = wald_test(sample2, theta, estimation_theta2, estimation_sigma2)

print('Уровень доверия: ', alpha / 2)
print('p-value для первой выборки: ', p_value1)
print('p-value для второй выборки: ', p_value2)
```

Обе отвергаем

```python
print(theta)
print('Доверительный интервал для первой выборки: ', conf_int1)
print('Доверительный интервал для второй выборки: ', conf_int2)
```

Наш предположительный параметр не попал в ни в один ДИ. Значит мы попали в точку, реализующуюся с вероятностью меньше 5 процентов. Причём в первый ДИ мы не попали чуть чуть, соотвествующий p-value не сильно меньше желаемого. Во второй мы сильно не попали, p-value совсем около нуля. Если же мы попадаем в ДИ, то гипотеза не отвергается. Такая взаимосвязь объясняется похожим определением для Критерий Вальда
