```python
# Скачиваем датасет из kaggle
path = kagglehub.dataset_download("camnugent/california-housing-prices")
data = pd.read_csv(os.path.join(path, "housing.csv"))
data = pd.get_dummies(data)
imputer = SimpleImputer(strategy='median')
data = pd.DataFrame(data=imputer.fit_transform(data), columns=data.columns)
data.head()
```

```python
X = data.drop(['median_house_value'], axis=1)
y = data['median_house_value']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=RANDOM_STATE)
```

```python
n_grid = np.arange(1, 401)
depth_grid = np.arange(1, 16)
n_mse_test = []
n_mse_train = []
depth_mse_test = []
depth_mse_train = []
```

```python
for n in tqdm(n_grid):
    model = GradientBoostingRegressor(n_estimators=n)
    model.fit(X_train, y_train)
    y_pred_test = model.predict(X_test)
    y_pred_train = model.predict(X_train)
    n_mse_test.append(mean_squared_error(y_pred_test, y_test))
    n_mse_train.append(mean_squared_error(y_pred_train, y_train))
```

```python
for depth in tqdm(depth_grid):
    model = GradientBoostingRegressor(max_depth=depth, n_estimators=400)
    model.fit(X_train, y_train)
    y_pred_test = model.predict(X_test)
    y_pred_train = model.predict(X_train)
    depth_mse_test.append(mean_squared_error(y_pred_test, y_test))
    depth_mse_train.append(mean_squared_error(y_pred_train, y_train))
```

```python
plt.figure(figsize=(8, 6))
plt.plot(n_grid, n_mse_train, label='MSE на тренировочной выборке')
plt.plot(n_grid, n_mse_test, label='MSE на тестовой выборке')
plt.title('Зависимость метрики MSE от количества деревьев')
plt.ylabel('MSE')
plt.xlabel('Количество деревьев')
plt.legend()
plt.show()

```

```python
plt.figure(figsize=(8, 6))
plt.plot(depth_grid, depth_mse_train, label='MSE на тренировочной выборке')
plt.plot(depth_grid, depth_mse_test, label='MSE на тестовой выборке')
plt.title('Зависимость метрики MSE от максимальной глубины дерева')
plt.ylabel('MSE ')
plt.xlabel('Максимальная глубина дерева')
plt.legend()
plt.show()
```

```python
lr_grid = np.linspace(0.01, 1, 15)
lr_mse_test = []
lr_mse_train = []
for lr in tqdm(lr_grid):
    model = GradientBoostingRegressor(n_estimators=400, max_depth=6, learning_rate=lr)
    model.fit(X_train, y_train)
    y_pred_test = model.predict(X_test)
    y_pred_train = model.predict(X_train)
    lr_mse_test.append(mean_squared_error(y_pred_test, y_test))
    lr_mse_train.append(mean_squared_error(y_pred_train, y_train))
```

```python
def calculate_asymptotic_confidence_intervals(
    sample: np.ndarray, alpha: float = 0.95
) -> Tuple[np.float64, np.float64]:
    """Вычисляет реализации статистик границ асимптотического доверительного интервала
    для параметра сдвига в нормальной модели в случае неизвестной дисперсии.

    Args:
        sample (np.ndarray): выборка
        alpha (float): уровень доверия
            (default is 0.95)

    Returns:
        left, right (Tuple[np.float64, np.float64]): левая и правая граница интервала
    """

    n = len(sample)
    x_bar = np.mean(sample)
    z = sps.norm.ppf((1 + alpha) / 2)
    s = np.std(sample)
    left = x_bar - z * s / np.sqrt(n)
    right = x_bar + z * s / np.sqrt(n)
    return (left, right)
```

```python
gb_left, gb_right = calculate_asymptotic_confidence_intervals(gb_times)
mean_val = np.mean(gb_times)
std_val = np.std(gb_times, ddof=1)
plt.figure(figsize=(8, 6))
sns.histplot(
    gb_times,
    bins=20,
    alpha=0.6,
    color='skyblue',
    label=f'Mean = {mean_val:.2e}, Std = {std_val:.2e}',
)
plt.axvline(mean_val, color='r', linestyle='--', label='Mean')
plt.axvline(gb_left, color='green', linestyle=':', label='CI lower')
plt.axvline(gb_right, color='green', linestyle=':', label='CI upper')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Frequency')
plt.title('Histogram of gb_times with CI')
plt.show()
```

```python
rf_left, rf_right = calculate_asymptotic_confidence_intervals(rf_times)
mean_val = np.mean(rf_times)
std_val = np.std(rf_times, ddof=1)
plt.figure(figsize=(8, 6))
sns.histplot(
    rf_times,
    bins=20,
    alpha=0.6,
    color='skyblue',
    label=f'Mean = {mean_val:.2e}, Std = {std_val:.2e}',
)
plt.axvline(mean_val, color='r', linestyle='--', label='Mean')
plt.axvline(rf_left, color='green', linestyle=':', label='CI lower')
plt.axvline(rf_right, color='green', linestyle=':', label='CI upper')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Frequency')
plt.title('Histogram of rf_times with CI')
plt.show()
```

```python
grid = np.linspace(imp_X_train.min(), imp_X_train.max(), 500).reshape(-1, 1)
staged = list(model.staged_predict(imp_X_train))
staged_grid = list(model.staged_predict(grid))

for i in range(10):
    if i == 0:
        previous_predict = 0
    else:
        previous_predict = staged[i - 1]

    error = y_train - previous_predict

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

    # Первый график
    scatter1 = ax1.scatter(imp_X_train, y_train, label="Обучающие данные", s=15, alpha=0.7)
    (line1,) = ax1.plot(
        grid, staged_grid[i], color="red", linewidth=2, label="Предсказание ансамбля"
    )
    ax1.set_xlabel("Признак")
    ax1.set_ylabel("Таргет")
    ax1.set_title(f"Итерация {i+1}")
    ax1.legend(fontsize=11, loc="best")

    # Второй график
    scatter2 = ax2.scatter(imp_X_train, error, label="Ошибка предыдущего шага", s=15, alpha=0.7)
    (line2,) = ax2.plot(
        grid,
        staged_grid[i] - (i > 0) * staged_grid[i - 1],
        color="red",
        linewidth=2,
        label=f"Коррекция {i+1}",
    )
    ax2.set_xlabel("Признак")
    ax2.set_ylabel("Ошибка")
    ax2.set_title(f"Итерация {i+1}")
    ax2.legend(fontsize=11, loc="best")

    plt.tight_layout()
    plt.show()
```

В данном случае это просто разность значений таргета и предсказаний на одном признаке. А вообще предсказания строятся на всей матрице признаков. Аналогичные картинки для каждого отдельного признака и ошибка - тоже векторная разница

**4.** **Функция лосса**

Для задачи регрессии разработано достаточно много функций потерь, в том числе с дополнительными свойствами робастности. Один такой пример — функция потерь Хьюбера, она же Huber loss. Суть функции в том, что на небольших отклонениях она работает как $\large L_2$, а с заранее заданного порога, начинает работать как $\large L_1$. Это позволяет уменьшить вклад выбросов и следующих за ними квадратично-больших ошибок на общий вид функции, при этом не акцентируя внимание на мелких неточностях и отклонениях.


$R(x)$ &mdash; функция потерь Хьюбера, определяемая как
$$R(x) = \frac{x^2}{2} I\left\{|x| < c\right\} + c \left(|x| - \frac{c}{2}\right)I\left\{|x| > c\right\}.$$

Обучите модель градиентного бустинга используя данный лосс сравните результат с MSE.

```python
model = GradientBoostingRegressor(n_estimators=n_estimators, max_depth=max_depth, learning_rate=learning_rate, loss='huber')
model.fit(X_train, y_train)
print(f'MSE при стандартном лоссе L2: {mean_squared_error(gb_model.predict(X_test), y_test)}')
print(f'MSE при лоссе Хьюбера: {mean_squared_error(model.predict(X_test), y_test)}')
```

**6.** Рассмотрим модели смеси градиентного бустинга $\widehat{y}_{gb}$ и случайного леса $\widehat{y}_{rf}$ с оптимальными гиперпараметрами, подобранными ранее, в виде
$$\widehat{y}(x) = w \widehat{y}_{gb}(x) +  (1-w) \widehat{y}_{rf}(x),$$
где $w \in [0, 1]$ &mdash; коэффициент усреднения. Подберите оптимальное значение гиперпараметра $w$ и выведите его. Удалось ли добиться улучшения качества на тестовой выборке?

```python
w_grid = np.linspace(0, 1, 30)
mses = []
for w in w_grid:
  y_gb = gb_model.predict(X_test)
  y_rf = rf_model.predict(X_test)
  mses.append(mean_squared_error(w * y_gb + (1 - w) * y_rf, y_test))

plt.figure(figsize=(8, 6))
plt.plot(w_grid, mses, color='skyblue')
plt.title('MSE(w)')
plt.xlabel('w')
plt.ylabel('MSE')
plt.show()
```

```python
w_min = w_grid[np.argmin(mses)]
print(f'Оптимальное w: {w_min:.2f}')
print(f'Наилучшая метрика: {np.min(mses):.2f}')
```

**Вывод:**

- Поскольку `w` не равен 0 или 1, **ни одна из моделей полностью не описывает закономерности** явления.  
- В целом, **Gradient Boosting (GB) быстрее обучается**, если не проводить тонкий подбор гиперпараметров, однако **подбор параметров сильно влияет на качество модели**.  
- GB строится **последовательно**, а не параллельно, корректируя свои предсказания на всем пространстве признаков **только по ошибкам предыдущих шагов**.
