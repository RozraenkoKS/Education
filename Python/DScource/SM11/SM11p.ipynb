**1.** Пусть $X = (X_1, \dots, X_n)$ &mdash; выборка из распределения $U[0, \theta]$. Рассмотрим класс оценок $\mathcal{K} = \left\{c_n X_{(n)}, c_n\in\mathbb{R}\right\}$. Выпишите формулы bias-variance разложения для таких оценок.

**Ответ:** 

Пусть $\hat\theta_{c_n} = c_n X_{(n)}$.
Тогда

$\mathrm{MSE}(c_n)

**2.** Пусть $X = (X_1, \dots, X_n)$ &mdash; выборка из распределения $\mathcal{N}(a, \sigma^2)$, где $a$ и $\sigma$ неизвестны. Для величины $\sigma^2$ рассмотрим класс оценок $\mathcal{K} = \left\{\frac{1}{c_n}\sum\limits_{i=1}^n \left(X_i - \overline{X}\right)^2, \; c_n \in\mathbb{R}\right\}$. Выпишите формулы bias-variance разложения для таких оценок.

**Ответ:** (см. лекцию)

Пусть $\hat\sigma^2_{c_n} = \frac{1}{c_n} \sum_{i=1}^n (X_i - \overline{X})^2$, где $X_i \sim \mathcal{N}(a, \sigma^2)$.  
Обозначим $S^2 = \sum_{i=1}^n (X_i - \overline{X})^2$.

- Математическое ожидание:
$$
\mathbb{E}[\hat\sigma^2_{c_n}] = \frac{1}{c_n} \mathbb{E}[S^2] = \frac{n-1}{c_n} \sigma^2
$$

- Смещение:
$$
\mathrm{Bias}(\hat\sigma^2_{c_n}) = \mathbb{E}[\hat\sigma^2_{c_n}] - \sigma^2 = \sigma^2\left(\frac{n-1}{c_n} - 1\right)
$$

- Дисперсия:
$$
\operatorname{Var}(\hat\sigma^2_{c_n}) = \frac{1}{c_n^2} \operatorname{Var}(S^2) = \frac{2 (n-1)}{c_n^2} \sigma^4
$$

- MSE:
$$
\mathrm{MSE}(\hat\sigma^2_{c_n}) = \operatorname{Var}(\hat\sigma^2_{c_n}) + \mathrm{Bias}^2(\hat\sigma^2_{c_n})
= \frac{2(n-1)}{c_n^2}\sigma^4 + \sigma^4 \left(\frac{n-1}{c_n}-1\right)^2
$$

**3.** Пусть $X = (X_1, \dots, X_n)$ &mdash; выборка из распределения $Exp (\theta)$. Рассмотрим класс оценок $\mathcal{K} = \left\{\frac{c}{X_1 + \dots + X_n}, \; c\in\mathbb{R}\right\}$. Выпишите формулы bias-variance разложения для таких оценок.






**Ответ:** (см. дз)

Пусть $\hat\theta_{c} = \frac{c}{\sum_{i=1}^n X_i}$, где $X_i \sim \mathrm{Exp}(\theta)$. Обозначим $S_n = \sum_{i=1}^n X_i$.  

- Математическое ожидание:
$$
\mathbb{E}[\hat\theta_c] = c \, \mathbb{E}\left[\frac{1}{S_n}\right]
$$
Напомним, что $S_n \sim \mathrm{Gamma}(n, \theta)$ с плотностью
$p_{S_n}(x) = \frac{\theta^n x^{n-1} e^{-\theta x}}{\Gamma(n)}, \; x>0$.

Тогда
$$
\mathbb{E}[\hat\theta_c] = c \int_0^\infty \frac{1}{x} \frac{\theta^n x^{n-1} e^{-\theta x}}{\Gamma(n)} dx
= c \frac{\theta^n}{\Gamma(n)} \int_0^\infty x^{n-2} e^{-\theta x} dx
= c \frac{\theta^n \Gamma(n-1)}{\Gamma(n) \theta^{n-1}} 
= c \frac{\theta}{n-1}, \quad n>1.
$$

- Смещение:
$$
\mathrm{Bias}(\hat\theta_c) = \mathbb{E}[\hat\theta_c] - \theta = \theta \left(\frac{c}{n-1} - 1\right)
$$

- Дисперсия:
$$
\operatorname{Var}(\hat\theta_c) = c^2 \operatorname{Var}\left(\frac{1}{S_n}\right)
= c^2 \left(\mathbb{E}\left[\frac{1}{S_n^2}\right] - \left(\mathbb{E}\frac{1}{S_n}\right)^2 \right)
$$
где
$$
\mathbb{E}\left[\frac{1}{S_n^2}\right] = \frac{\theta^2}{(n-1)^2(n-2)}, \quad n>2
$$

- MSE:
$$
\mathrm{MSE}(\hat\theta_c) = \operatorname{Var}(\hat\theta_c) + \mathrm{Bias}^2(\hat\theta_c)
= c^2 \frac{\theta^2}{(n-1)^2(n-2)} - c^2 \frac{\theta^2}{(n-1)^2} + \theta^2 \left(\frac{c}{n-1}-1\right)^2
= \theta^2 \left[ \frac{c^2}{(n-2)(n-1)} + \left(\frac{c}{n-1}-1\right)^2 \right], \quad n>2
$$




Повторите исследование, аналогичное пункту *1* для $\theta = 1$ и $n = 7$. Не забудьте сделать выводы.

**Решение:**

```python
theta = 1
n = 7
c = np.linspace(0.5, 10, 2000)

mse = theta**2 * (c**2 / ((n-1)**2*(n-2)) + (c/(n-1) - 1)**2)
bias2 = theta**2 * (c/(n-1) - 1)**2
var = theta**2 * (c**2 / ((n-1)**2*(n-2)))

c_opt = n - 2
c_unb = n - 1
c_mle = n

plt.figure(figsize=(8,5))
plt.plot(c, mse, label=r'$\mathrm{MSE}(c)$', linewidth=2)

plt.axvline(c_opt, color='C1', linestyle='--',
            label=rf'Оптимум MSE: $c^*={c_opt:.3f}$')
plt.axvline(c_unb, color='C2', linestyle='--',
            label=rf'Несмещённая: $c_{{unb}}={c_unb}$')
plt.axvline(c_mle, color='C3', linestyle='--',
            label=rf'MLE: $c_{{mle}}={c_mle}$')

plt.xlim(0.5, 10)
plt.xlabel(r'Коэффициент $c$')
plt.ylabel(r'$\mathrm{MSE}(c)$')
plt.title(r'MSE для оценки $c / \sum X_i$ при $n=7,\,\theta=1$')
plt.grid(alpha=0.3)
plt.legend(fontsize=12, bbox_to_anchor=(1,1))
plt.show()

plt.figure(figsize=(8,5))
plt.plot(c, bias2, label=r'$\mathrm{Bias}^2(c)$', linewidth=2)

plt.axvline(c_opt, color='C1', linestyle='--')
plt.axvline(c_unb, color='C2', linestyle='--')
plt.axvline(c_mle, color='C3', linestyle='--')

plt.xlim(0.5, 10)
plt.xlabel(r'Коэффициент $c$')
plt.ylabel(r'$\mathrm{Bias}^2(c)$')
plt.title(r'Квадрат смещения, $n=7$')
plt.grid(alpha=0.3)
plt.legend(fontsize=12, bbox_to_anchor=(1,1))
plt.show()

plt.figure(figsize=(8,5))
plt.plot(c, var, label=r'$\mathrm{Var}(c)$', linewidth=2)

plt.axvline(c_opt, color='C1', linestyle='--')
plt.axvline(c_unb, color='C2', linestyle='--')
plt.axvline(c_mle, color='C3', linestyle='--')

plt.xlim(0.5, 10)
plt.xlabel(r'Коэффициент $c$')
plt.ylabel(r'$\mathrm{Var}(c)$')
plt.title(r'Дисперсия, $n=7$')
plt.grid(alpha=0.3)
plt.legend(fontsize=12, bbox_to_anchor=(1,1))
plt.show()
```

### Задача 2.

где $L$ — одна из двух функций потерь: квадратичная $L(x, y)=(x-y)^2$ и линейная $L(x, y)=\left|x-y\right|$.

Для каждого из типов функций потерь постройте свой график. Нанесите на этот график для каждой из четырех оценок $\widehat{\theta}$ оценку функции потерь $\widehat{R}_{\widehat{\theta}}(\theta)$. Ограничение сверху по оси *y* ставьте таким, чтобы графики функции риска с малыми значениями четко различались.

*Совет*: при тестировании кода запускайте его с небольшими размерами данных. Например, используйте 100 реализаций выборок. Финальные результаты получите, поставив требуемые значения размеров данных.

В задаче доспустимо использовать короткие циклы, то есть с малым количествои итераций.

**Решение:**

```python
theta_grid = np.arange(0.01, 2 + 0.01, 0.01)
r1, r2, r3, r4 = [], [], [], []

for theta in tqdm(theta_grid):
    x = sps.uniform(0, theta).rvs(size=(5000, 100))
    
    est1 = 2 * np.mean(x, axis=1)
    est2 = 101 * np.min(x, axis=1)
    est3 = np.min(x, axis=1) + np.max(x, axis=1)
    est4 = 101 / 100 * np.max(x, axis=1)
    
    l1_1 = np.sum((est1 - theta) ** 2) / 5000
    l1_2 = np.sum((est2 - theta) ** 2) / 5000
    l1_3 = np.sum((est3 - theta) ** 2) / 5000
    l1_4 = np.sum((est4 - theta) ** 2) / 5000

    r1.append(l1_1)
    r2.append(l1_2)
    r3.append(l1_3)
    r4.append(l1_4)
```

```python
plt.figure(figsize=(10,6))

plt.plot(theta_grid, r1, label=r'$2\bar X$', linewidth=2)
plt.plot(theta_grid, r2, label=r'$(n+1)\min X$', linewidth=2)
plt.plot(theta_grid, r3, label=r'$\min X + \max X$', linewidth=2)
plt.plot(theta_grid, r4, label=r'$\frac{n+1}{n}\max X$', linewidth=2)

plt.xlabel(r'$\theta$', fontsize=12)
plt.ylabel(r'Эмпирический $L_1$ риск', fontsize=12)
plt.title(r'$L_1$–риск оценок параметра $\theta$ для $U(0,\theta)$', fontsize=14)

plt.ylim(0, 0.005)
plt.grid(alpha=0.3)
plt.legend(fontsize=12, bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
